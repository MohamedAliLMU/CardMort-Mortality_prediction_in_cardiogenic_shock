```python
import seaborn as sns
import matplotlib.pyplot as plt

# Stil setzen
sns.set(style="whitegrid", font_scale=1.1)

def plot_metrics_stacked(metrics_df):
    fig, axes = plt.subplots(2, 1, figsize=(10, 8), sharex=True)

    # ROC AUC Plot (oben)
    sns.boxplot(
        x="Group", 
        y="ROC AUC", 
        hue="Group By", 
        data=metrics_df,
        palette="Set2",
        ax=axes[0]
    )
    axes[0].set_title("ROC AUC by Group (eICU, tested with models from different datasets)")
    axes[0].set_xlabel("")
    axes[0].set_ylabel("ROC AUC")
    axes[0].tick_params(axis='x', rotation=45)
    axes[0].legend_.remove()  # nur eine Legende unten

    # PR AUC Plot (unten)
    sns.boxplot(
        x="Group", 
        y="PR AUC", 
        hue="Group By", 
        data=metrics_df,
        palette="Set2",
        ax=axes[1]
    )
    axes[1].set_title("PR AUC by Group (eICU, tested with models from different datasets)")
    axes[1].set_xlabel("Group")
    axes[1].set_ylabel("PR AUC")
    axes[1].tick_params(axis='x', rotation=45)
    axes[1].legend_.remove()  # nur eine Legende unten

    # Gemeinsame Legende unten zentriert
    handles, labels = axes[1].get_legend_handles_labels()
    fig.legend(handles, labels, loc='lower center', bbox_to_anchor=(0.5, -0.02), ncol=3)

    plt.tight_layout(rect=[0, 0.03, 1, 0.97])
    plt.show()

# Aufruf der Funktion
plot_metrics_stacked(metrics_df)
```
